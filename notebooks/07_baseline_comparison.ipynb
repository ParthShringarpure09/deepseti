```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline 
```

```python
overall_results = pd.DataFrame({
    "model": [
        "PCA",
        "ConvAutoencoderV3",
    ],
    "AUROC": [
        0.6768551020408163,
        0.6958867346938775,
    ],
    "AUPRC": [
        0.6435536752089339,
        0.6705543876345372,
    ],
})

overall_results
```

```python
plot_df = overall_results.set_index("model")

ax = plot_df.plot(
    kind="bar",
    figsize=(8, 5),
)

ax.set_title(
    "DeepSeti Baseline Anomaly Detection Performance"
)

ax.set_ylabel("Score")
ax.set_xlabel("Model")

ax.set_ylim(0.0, 1.0)

plt.xticks(
    rotation=0,
)

plt.legend(
    title="Metric",
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(
    figsize=(8, 5)
)

plt.plot(
    snr_results["snr"],
    snr_results["pca_auroc"],
    marker="o",
    label="PCA",
)

plt.plot(
    snr_results["snr"],
    snr_results["convae_auroc"],
    marker="o",
    label="ConvAutoencoderV3",
)

plt.axhline(
    0.5,
    linestyle="--",
    label="Random baseline",
)

plt.xlabel("Injected signal SNR")
plt.ylabel("AUROC")

plt.title(
    "Anomaly Detection Performance vs Signal Strength"
)

plt.ylim(
    0.45,
    1.0,
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
