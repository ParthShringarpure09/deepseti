```python
# 1. path setup
import sys
from pathlib import Path
repo_root = Path.cwd()
while not (repo_root / "src").exists():
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root))
```

```python
# 2. imports + gather files + split
from torch.utils.data import DataLoader
from src.data.dataset import SpectrogramDataset, split_files

quiet_dir = repo_root / "data/synthetic/quiet"
real_dir  = repo_root / "data/processed/real"
all_files = sorted(quiet_dir.glob("*.npy")) + sorted(real_dir.glob("*.npy"))
train_files, val_files, test_files = split_files(all_files)
print("train/val/test:", len(train_files), len(val_files), len(test_files))
```

```python
# 3. datasets + loaders + sanity check
train_ds = SpectrogramDataset(files=train_files)
val_ds   = SpectrogramDataset(files=val_files)
test_ds  = SpectrogramDataset(files=test_files)

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=16, shuffle=False)
test_loader  = DataLoader(test_ds, batch_size=16, shuffle=False)

batch = next(iter(train_loader))
print("batch shape:", batch.shape)
print("mean/std:", batch.mean().item(), batch.std().item())
```
