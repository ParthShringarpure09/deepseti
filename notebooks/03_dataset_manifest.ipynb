```python
from pathlib import Path

from deepseti.data import manifest

PROJECT_ROOT = Path.cwd().parent
RAW_ROOT = PROJECT_ROOT / "data" / "raw"

rows = manifest.load_manifest(
    PROJECT_ROOT / "configs" / "data" / "observations.csv"
)

print("Loaded rows:", len(rows))
```

```python
from blimpy import Waterfall

for row in rows:
    file_path = (
        RAW_ROOT
        / row["raw_subdir"]
        / row["filename"]
    )

    try:
        wf = Waterfall(str(file_path), load_data=False)

        print(
            row["observation_id"],
            "| source:", wf.header["source_name"],
            "| nchans:", wf.header["nchans"],
            "| tsamp:", wf.header["tsamp"],
            "| OK",
        )

    except Exception as error:
        print(
            row["observation_id"],
            "| ERROR:",
            error,
        )
```
