```python
import numpy as np
from pathlib import Path

def generate_quiet_batch(n_frames: int, out_dir: str = "data/synthetic/quiet", seed_start: int = 0):
    out_path = Path(out_dir)
    out_path.mkdir(parents=True, exist_ok=True)

    for i in range(n_frames):
        frame = generate_quiet_frame(seed=seed_start + i)
        np.save(out_path / f"quiet_{i:05d}.npy", frame)

    print(f"Saved {n_frames} frames to {out_path}")
```

```python
generate_quiet_batch(n_frames=5, out_dir=str(repo_root / "data/synthetic/quiet"))
```
